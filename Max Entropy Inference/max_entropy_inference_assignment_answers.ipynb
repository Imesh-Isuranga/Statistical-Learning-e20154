# Q. Statistical Mechanics of the Quantum Oscillator

Consider a single quantum harmonic oscillator of angular frequency $\omega$ in thermal equilibrium with a heat bath at temperature $T$. The quantum mechanical energy eigenvalues are given by:

$$E_j = \hbar \omega \left(j + \frac{1}{2}\right), \quad \text{for } j = 0, 1, 2, \dots$$

where $\hbar$ is the reduced Planck constant and $k_B$ is the Boltzmann constant.

**Answer the Following:**

1. **The Partition Function:**
Show that the canonical partition function $Z(T) = \sum_{j=0}^\infty e^{-E_j / (k_B T)}$ forms a convergent geometric series that simplifies to:

$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$


2. **Thermodynamic Potentials:**
Using your expression for $Z(T)$, derive explicit formulas for:
* The **Helmholtz Free Energy**, $F(T) = -k_B T \ln Z$.
* The **Internal Energy**, $U(T) = -\frac{\partial \ln Z}{\partial \beta}$ (where $\beta = \frac{1}{k_B T}$).


3. **Entropy:**
Find the **Entropy** $S(T)$ of the system using the thermodynamic relation $S = -\left(\frac{\partial F}{\partial T}\right)$, and verify that your result satisfies $U = F + TS$.
4. **Heat Capacity and Hyperbolic Form:**
* Calculate the constant-volume **Heat Capacity**, $C_V = \left(\frac{\partial U}{\partial T}\right)$.
* Use the hyperbolic identity $\sinh(x) = \frac{e^x - e^{-x}}{2}$ to prove that your expression for $C_V$ can be compactly rewritten as:

$$C_V = k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$




5. **Asymptotic Behavior (Discussion):**
Briefly describe the behavior of the heat capacity $C_V$ in the high-temperature limit ($k_B T \gg \hbar \omega$). Does it match the classical equipartition theorem prediction?

# Answer

## Q1. The Partition Function

The canonical partition function is

$$
Z(T) = \sum_{j=0}^{\infty} e^{-E_j/(k_BT)}, \qquad E_j = \hbar\omega\left(j+\tfrac12\right)
$$

Substituting $E_j$:

$$
Z(T) = \sum_{j=0}^{\infty} e^{-\beta \hbar\omega (j+1/2)}
= e^{-\beta\hbar\omega/2}\sum_{j=0}^{\infty} \left(e^{-\beta\hbar\omega}\right)^{j},
\qquad \beta \equiv \frac{1}{k_BT}
$$

This is a **geometric series** with ratio $r = e^{-\beta\hbar\omega}$. Since $\omega>0, T>0$, we have $0<r<1$, so the series converges:

$$
\sum_{j=0}^{\infty} r^j = \frac{1}{1-r}
$$

Therefore:

$$
\boxed{Z(T) = \frac{e^{-\hbar\omega/(2k_BT)}}{1-e^{-\hbar\omega/(k_BT)}}}
$$

## Q2. Thermodynamic Potentials

### Log of the partition function

It is easiest to work with $\ln Z$ first:

$$
\ln Z(T) = -\frac{\hbar\omega}{2k_BT} - \ln\left(1-e^{-\hbar\omega/(k_BT)}\right)
$$

### Helmholtz Free Energy

$$
F(T) = -k_BT\ln Z = -k_BT\left[-\frac{\hbar\omega}{2k_BT} - \ln\left(1-e^{-\hbar\omega/(k_BT)}\right)\right]
$$

$$
\boxed{F(T) = \frac{\hbar\omega}{2} + k_BT\ln\left(1-e^{-\hbar\omega/(k_BT)}\right)}
$$

### Internal Energy

Using $\beta = 1/(k_BT)$, write $\ln Z$ in terms of $\beta$:

$$
\ln Z(\beta) = -\frac{\beta\hbar\omega}{2} - \ln\left(1-e^{-\beta\hbar\omega}\right)
$$

Then:

$$
U = -\frac{\partial \ln Z}{\partial \beta}
= \frac{\hbar\omega}{2} - \frac{\partial}{\partial\beta}\left[-\ln(1-e^{-\beta\hbar\omega})\right]
$$

Compute the derivative:

$$
\frac{\partial}{\partial \beta}\left[-\ln(1-e^{-\beta\hbar\omega})\right]
= -\frac{\hbar\omega\, e^{-\beta\hbar\omega}}{1-e^{-\beta\hbar\omega}}
$$

So:

$$
U = \frac{\hbar\omega}{2} + \frac{\hbar\omega\, e^{-\beta\hbar\omega}}{1-e^{-\beta\hbar\omega}}
$$

Multiplying numerator and denominator by $e^{\beta\hbar\omega/2}$ gives the standard form:

$$
\boxed{U(T) = \frac{\hbar\omega}{2} + \frac{\hbar\omega}{e^{\hbar\omega/(k_BT)}-1}}
$$

This is the famous **Planck oscillator energy**: zero-point energy $\hbar\omega/2$ plus $\hbar\omega \bar n(T)$, where $\bar n = 1/(e^{\beta\hbar\omega}-1)$ is the Bose–Einstein occupation number.

## Q3. Entropy

We have

$$
F(T) = \frac{\hbar\omega}{2} + k_BT\ln\left(1-e^{-\hbar\omega/(k_BT)}\right)
$$

Entropy is $S = -\left(\dfrac{\partial F}{\partial T}\right)_V$. Let $x \equiv \dfrac{\hbar\omega}{k_BT}$, so $F = \dfrac{\hbar\omega}{2} + k_BT\ln(1-e^{-x})$.

Differentiate term by term (product rule on the second term):

$$
\frac{\partial F}{\partial T} = k_B\ln(1-e^{-x}) + k_BT \cdot \frac{\partial}{\partial T}\ln(1-e^{-x})
$$

Since $x = \hbar\omega/(k_BT)$, $\dfrac{\partial x}{\partial T} = -\dfrac{\hbar\omega}{k_BT^2} = -\dfrac{x}{T}$, hence:

$$
\frac{\partial}{\partial T}\ln(1-e^{-x}) = \frac{e^{-x}}{1-e^{-x}}\cdot\left(-\frac{\partial x}{\partial T}\right)\cdot(-1)^{?}
$$

More directly:

$$
\frac{\partial}{\partial T}\ln(1-e^{-x}) = \frac{1}{1-e^{-x}}\cdot e^{-x}\cdot\frac{x}{T}
$$

So:

$$
\frac{\partial F}{\partial T} = k_B\ln(1-e^{-x}) + k_B\, x\,\frac{e^{-x}}{1-e^{-x}}
$$

Therefore:

$$
S = -\frac{\partial F}{\partial T} = -k_B\ln(1-e^{-x}) - k_B\,\frac{x\,e^{-x}}{1-e^{-x}}
$$

Rewriting $x e^{-x}/(1-e^{-x}) = x/(e^{x}-1)$:

$$
\boxed{S(T) = k_B\left[\frac{x}{e^{x}-1} - \ln\left(1-e^{-x}\right)\right], \qquad x=\frac{\hbar\omega}{k_BT}}
$$

or equivalently, in original variables:

$$
S(T) = \frac{\hbar\omega/T}{e^{\hbar\omega/(k_BT)}-1} - k_B\ln\left(1-e^{-\hbar\omega/(k_BT)}\right)
$$

### Verification: $U = F + TS$

$$
F + TS = \left[\frac{\hbar\omega}{2}+k_BT\ln(1-e^{-x})\right] + T\left[\frac{k_B x}{e^x-1} - k_B\ln(1-e^{-x})\right]
$$

The $k_BT\ln(1-e^{-x})$ terms cancel exactly:

$$
F+TS = \frac{\hbar\omega}{2} + k_BT\cdot\frac{x}{e^x-1} = \frac{\hbar\omega}{2} + k_BT\cdot\frac{\hbar\omega/(k_BT)}{e^x-1} = \frac{\hbar\omega}{2}+\frac{\hbar\omega}{e^x-1}
$$

which is exactly $U(T)$ found in Q2. $\checkmark$

## Q4. Heat Capacity and Hyperbolic Form

### Direct differentiation

Starting from

$$
U(T) = \frac{\hbar\omega}{2} + \frac{\hbar\omega}{e^{\hbar\omega/(k_BT)}-1}
$$

let $x = \hbar\omega/(k_BT)$. Then

$$
C_V = \frac{\partial U}{\partial T} = \hbar\omega \cdot \frac{\partial}{\partial T}\left[\frac{1}{e^x-1}\right]
$$

By the chain rule,

$$
\frac{\partial}{\partial T}\left[\frac{1}{e^x-1}\right] = -\frac{e^x}{(e^x-1)^2}\cdot\frac{\partial x}{\partial T}
= -\frac{e^x}{(e^x-1)^2}\cdot\left(-\frac{x}{T}\right) = \frac{x\,e^x}{T(e^x-1)^2}
$$

So:

$$
C_V = \hbar\omega \cdot \frac{x\, e^x}{T(e^x-1)^2} = \frac{\hbar\omega\, x}{T}\cdot\frac{e^x}{(e^x-1)^2}
$$

Since $\hbar\omega x/T = \hbar\omega \cdot \dfrac{\hbar\omega}{k_BT}\cdot\dfrac1T = k_B x^2$ (using $x=\hbar\omega/(k_BT)$):

$$
\boxed{C_V = k_B x^2 \frac{e^{x}}{(e^{x}-1)^2}, \qquad x = \frac{\hbar\omega}{k_BT}}
$$

### Converting to hyperbolic form

Factor $e^{x/2}$ from numerator and denominator:

$$
\frac{e^x}{(e^x-1)^2} = \frac{e^x}{\left(e^{x/2}(e^{x/2}-e^{-x/2})\right)^2} = \frac{e^x}{e^{x}\left(e^{x/2}-e^{-x/2}\right)^2} = \frac{1}{\left(e^{x/2}-e^{-x/2}\right)^2}
$$

Using $\sinh(x/2) = \dfrac{e^{x/2}-e^{-x/2}}{2}$, i.e. $e^{x/2}-e^{-x/2} = 2\sinh(x/2)$:

$$
\frac{e^x}{(e^x-1)^2} = \frac{1}{4\sinh^2(x/2)}
$$

Substituting back:

$$
C_V = k_B x^2 \cdot \frac{1}{4\sinh^2(x/2)} = k_B\left(\frac{x}{2}\right)^2\frac{1}{\sinh^2(x/2)}
$$

Recalling $x = \hbar\omega/(k_BT)$, so $x/2 = \hbar\omega/(2k_BT)$:

$$
\boxed{C_V = k_B\left(\frac{\hbar\omega}{2k_BT}\right)^2\frac{1}{\sinh^2\left(\dfrac{\hbar\omega}{2k_BT}\right)}}
$$

which matches the required form exactly. $\checkmark$

## Q5. High-Temperature Limit (Discussion)

In the limit $k_BT \gg \hbar\omega$, i.e. $x = \hbar\omega/(k_BT) \to 0$, expand $\sinh$:

$$
\sinh\left(\frac{x}{2}\right) \approx \frac{x}{2} + \frac{1}{6}\left(\frac{x}{2}\right)^3 + \cdots \approx \frac{x}{2}
$$

so

$$
\sinh^2\left(\frac{x}{2}\right) \approx \frac{x^2}{4}
$$

Substituting into $C_V$:

$$
C_V \approx k_B\left(\frac{x}{2}\right)^2 \cdot \frac{4}{x^2} = k_B
$$

**Result:** as $T\to\infty$ (or more precisely $k_BT\gg \hbar\omega$),

$$
\boxed{C_V \to k_B}
$$

This **exactly matches the classical equipartition theorem**: a 1‑D harmonic oscillator has two quadratic degrees of freedom (kinetic $p^2/2m$ and potential $\tfrac12 m\omega^2 x^2$), each contributing $\tfrac12 k_B$ to the heat capacity, giving $C_V = \tfrac12k_B+\tfrac12k_B=k_B$.

**Physical picture:** at high temperature, $k_BT$ is much larger than the energy-level spacing $\hbar\omega$, so the discreteness of the spectrum becomes irrelevant — the oscillator behaves as if its energy levels form a continuum, recovering classical statistical mechanics. Quantum effects (and the suppression of $C_V$) only become important at low temperature, $k_BT \lesssim \hbar\omega$, where $C_V \to 0$ exponentially (a signature of the "freezing out" of the oscillator's degree of freedom — this is precisely what resolved the classical specific-heat puzzle of solids, e.g. the Einstein solid model).

```python
import sympy as sp

# Symbols
hbar, omega, kB, T, beta, x = sp.symbols('hbar omega k_B T beta x', positive=True)
j = sp.symbols('j', integer=True, nonnegative=True)

# --- Q1: Partition function as a sum, verified in closed form ---
E_j = hbar*omega*(j + sp.Rational(1,2))
Z_sum = sp.summation(sp.exp(-E_j/(kB*T)), (j, 0, sp.oo))
Z_closed = sp.exp(-hbar*omega/(2*kB*T)) / (1 - sp.exp(-hbar*omega/(kB*T)))

print("Z from direct summation :", sp.simplify(Z_sum))
print("Z closed-form           :", sp.simplify(Z_closed))
print("Match:", sp.simplify(Z_sum - Z_closed) == 0)

# --- Q2: Free energy and Internal energy ---
lnZ = sp.log(Z_closed)
F = -kB*T*lnZ
F_simplified = sp.simplify(sp.expand_log(F, force=True))
print("\nF(T) =", F_simplified)

lnZ_beta = lnZ.subs(T, 1/(kB*beta))   # express in terms of beta
U = -sp.diff(lnZ_beta, beta)
U_simplified = sp.simplify(U).subs(beta, 1/(kB*T))
U_simplified = sp.simplify(U_simplified)
print("U(T) =", U_simplified)

# --- Q3: Entropy, and check U = F + T*S ---
S = sp.simplify(-sp.diff(F_simplified, T))
print("\nS(T) =", S)

check_UFS = sp.simplify(U_simplified - (F_simplified + T*S))
print("U - (F + T*S) simplifies to (should be 0):", check_UFS)

# --- Q4: Heat capacity, direct and hyperbolic form ---
Cv = sp.simplify(sp.diff(U_simplified, T))
print("\nC_V (direct) =", Cv)

x_val = hbar*omega/(2*kB*T)
Cv_hyperbolic = kB * x_val**2 / sp.sinh(x_val)**2

diff_check = sp.simplify(Cv - Cv_hyperbolic)
print("C_V direct - C_V hyperbolic (should be 0):", diff_check)

# --- Q5: High-T limit ---
Cv_limit = sp.limit(Cv_hyperbolic, T, sp.oo)
print("\nlim_{T->inf} C_V =", Cv_limit, " (should equal k_B)")

# Numeric sanity check with actual constants
import numpy as np
hbar_val, kB_val, omega_val = 1.0545718e-34, 1.380649e-23, 1e13  # rad/s
T_vals = np.logspace(0, 4, 5)
Cv_numeric = [float(Cv_hyperbolic.subs({hbar:hbar_val, kB:kB_val, omega:omega_val, T:t})) for t in T_vals]
for t, c in zip(T_vals, Cv_numeric):
    print(f"T={t:10.2f} K   C_V={c:.4e} J/K   (k_B={kB_val:.4e})")
```

# Q. The Maximum Entropy Distribution under Mean and Variance Constraints

Consider the random variable $X \in (-\infty, \infty)$. Let:

1. **The Mean:** $\mathbb{E}[X] = \mu$
2. **The Variance:** $\mathbb{E}[(X-\mu)^2] = \sigma^2$

Show that the law of $X$ that maximizes entropy is given by:

$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

# Answer

## Maximum Entropy Distribution under Mean and Variance Constraints

We seek a probability density $p(x)$ on $x\in(-\infty,\infty)$ that **maximizes the differential (Shannon) entropy**

$$
H[p] = -\int_{-\infty}^{\infty} p(x)\ln p(x)\,dx
$$

subject to three constraints:

$$
\text{(C0) Normalization:} \quad \int_{-\infty}^{\infty} p(x)\,dx = 1
$$

$$
\text{(C1) Mean:} \quad \int_{-\infty}^{\infty} x\,p(x)\,dx = \mu
$$

$$
\text{(C2) Variance:} \quad \int_{-\infty}^{\infty} (x-\mu)^2\,p(x)\,dx = \sigma^2
$$

We solve this using the method of **Lagrange multipliers in a functional (calculus of variations) setting**.

### Step 1: Construct the Constrained Functional

Introduce Lagrange multipliers $\lambda_0, \lambda_1, \lambda_2$ for constraints (C0), (C1), (C2) respectively, and form the functional

$$
\mathcal{L}[p] = -\int p(x)\ln p(x)\,dx
+ \lambda_0\left(\int p(x)\,dx - 1\right)
+ \lambda_1\left(\int x\,p(x)\,dx - \mu\right)
+ \lambda_2\left(\int (x-\mu)^2 p(x)\,dx - \sigma^2\right)
$$

This can be written as a single integral over an effective Lagrangian density $\mathcal{F}(x,p)$:

$$
\mathcal{L}[p] = \int_{-\infty}^{\infty}\Big[-p\ln p + \lambda_0\, p + \lambda_1\, x\,p + \lambda_2 (x-\mu)^2 p\Big]dx \;-\;\lambda_0 - \lambda_1\mu - \lambda_2\sigma^2
$$

### Step 2: Functional Derivative (Euler–Lagrange Condition)

We require $\mathcal{L}[p]$ to be **stationary** with respect to arbitrary variations $p(x)\to p(x)+\epsilon\,\eta(x)$. Since the integrand $\mathcal{F}$ depends on $p$ but not $p'$, the Euler–Lagrange equation reduces to simply setting the partial derivative with respect to $p$ to zero:

$$
\frac{\partial \mathcal{F}}{\partial p} = 0
\quad\Longrightarrow\quad
\frac{\partial}{\partial p}\Big[-p\ln p + \lambda_0 p + \lambda_1 x p + \lambda_2 (x-\mu)^2 p\Big] = 0
$$

Differentiating term by term (recall $\dfrac{d}{dp}(p\ln p) = \ln p + 1$):

$$
-(\ln p + 1) + \lambda_0 + \lambda_1 x + \lambda_2 (x-\mu)^2 = 0
$$

### Step 3: Solve for $p(x)$

Rearranging the stationarity condition:

$$
\ln p(x) = \lambda_0 - 1 + \lambda_1 x + \lambda_2 (x-\mu)^2
$$

Exponentiating both sides:

$$
p(x) = \exp\left[\lambda_0 - 1 + \lambda_1 x + \lambda_2(x-\mu)^2\right]
$$

This is the **general exponential-family form** dictated by MaxEnt: an exponential of a linear combination of the constraint functions ($1$, $x$, and $(x-\mu)^2$). All that remains is to **fix the multipliers** $\lambda_0,\lambda_1,\lambda_2$ using the three constraints.

### Step 4: Fixing $\lambda_1$ using the Mean Constraint

By symmetry of the ansatz, it is convenient to complete the square. Write the exponent as a quadratic in $x$:

$$
\lambda_1 x + \lambda_2(x-\mu)^2 = \lambda_2 x^2 + (\lambda_1 - 2\lambda_2\mu)x + \lambda_2\mu^2
$$

For $p(x)$ to be a normalizable density on $(-\infty,\infty)$, we need $\lambda_2 < 0$ (so the quadratic form is negative and $p\to 0$ as $|x|\to\infty$). Complete the square:

$$
\lambda_2 x^2 + (\lambda_1-2\lambda_2\mu)x = \lambda_2\left(x - \frac{2\lambda_2\mu - \lambda_1}{2\lambda_2}\right)^2 - \frac{(\lambda_1-2\lambda_2\mu)^2}{4\lambda_2}
$$

For the resulting distribution to be **centered exactly at $\mu$** (as required by constraint C1, $\mathbb{E}[X]=\mu$), the linear term must vanish independently, i.e.

$$
\lambda_1 - 2\lambda_2\mu = 0 \quad\Longrightarrow\quad \lambda_1 = 2\lambda_2\mu
$$

Substituting back, the exponent simplifies beautifully to:

$$
\lambda_1 x + \lambda_2(x-\mu)^2 = \lambda_2 x^2 - 2\lambda_2\mu x + \lambda_2\mu^2 = \lambda_2(x-\mu)^2
$$

(This is consistent — it says the cross term cancels and we are left purely with a $(x-\mu)^2$ dependence, confirming the natural choice $\lambda_1=2\lambda_2\mu$ makes the mean constraint automatically compatible with symmetry about $\mu$.)

So the density takes the reduced form:

$$
p(x) = A\,\exp\left[\lambda_2 (x-\mu)^2\right], \qquad A \equiv e^{\lambda_0 - 1}
$$

Define $\lambda_2 \equiv -\dfrac{1}{2\sigma^2}$ (anticipating the variance constraint; the negative sign ensures normalizability). Then:

$$
p(x) = A\exp\left[-\frac{(x-\mu)^2}{2\sigma^2}\right]
$$

### Step 5: Fix $A$ using Normalization (C0)

We require

$$
\int_{-\infty}^{\infty} A\,e^{-(x-\mu)^2/(2\sigma^2)}\,dx = 1
$$

Using the standard **Gaussian integral** $\displaystyle\int_{-\infty}^{\infty} e^{-t^2/(2a^2)}\,dt = a\sqrt{2\pi}$ (substitute $t=x-\mu$, $a=\sigma$):

$$
A\cdot \sigma\sqrt{2\pi} = 1 \quad\Longrightarrow\quad A = \frac{1}{\sqrt{2\pi\sigma^2}}
$$

Therefore, the maximum entropy distribution is:

$$
\boxed{p(x) = \frac{1}{\sqrt{2\pi\sigma^2}}\exp\left(-\frac{(x-\mu)^2}{2\sigma^2}\right)}
$$

which is exactly the **Normal (Gaussian) distribution** $\mathcal{N}(\mu,\sigma^2)$.

### Step 6: Verify the Constraints Are Satisfied

**Mean:** By symmetry of $p(x)$ about $x=\mu$ (the exponent depends only on $(x-\mu)^2$), the density is an even function of $(x-\mu)$, so

$$
\mathbb{E}[X] = \int_{-\infty}^{\infty} x\,p(x)\,dx = \mu \quad \checkmark
$$

**Variance:** Substituting $t = x-\mu$,

$$
\mathbb{E}[(X-\mu)^2] = \int_{-\infty}^{\infty} t^2 \cdot \frac{1}{\sqrt{2\pi\sigma^2}} e^{-t^2/(2\sigma^2)}\,dt = \sigma^2 \quad \checkmark
$$

(this is the second moment of the standard Gaussian integral, a well-known identity obtained via integration by parts or differentiating the Gaussian integral with respect to a parameter).

### Step 7: Confirm This Is a Maximum (Not Just a Stationary Point)

The entropy functional $H[p]=-\int p\ln p\,dx$ is **strictly concave** in $p$ (since $-p\ln p$ is a strictly concave function of $p$ for $p>0$, and the constraint set is convex — it is an intersection of affine/linear constraints on $p$). Therefore, the unique stationary point found via Lagrange multipliers is guaranteed to be the **global maximum** of entropy subject to the constraints, not merely a saddle point or local extremum.

### Interpretation

This result is the continuous analogue of the discrete MaxEnt principle: **among all probability distributions on $\mathbb{R}$ with a specified mean and variance, the Gaussian distribution has the maximum (differential) entropy.** No other assumption about the "shape" of $X$ is needed — Gaussianity emerges purely from the requirement of maximum uncertainty (least additional structure) consistent with knowing only the first two moments.

### Entropy of the Maximum Entropy Gaussian

It's instructive to compute $H[p]$ at the optimum. Using $\ln p(x) = -\ln\sqrt{2\pi\sigma^2} - \dfrac{(x-\mu)^2}{2\sigma^2}$:

$$
H[p] = -\mathbb{E}[\ln p(X)] = \ln\sqrt{2\pi\sigma^2} + \frac{\mathbb{E}[(X-\mu)^2]}{2\sigma^2}
$$

Since $\mathbb{E}[(X-\mu)^2] = \sigma^2$ (constraint C2):

$$
H[p] = \ln\sqrt{2\pi\sigma^2} + \frac{1}{2} = \frac{1}{2}\ln(2\pi\sigma^2) + \frac{1}{2}
$$

$$
\boxed{H[p_{\max}] = \frac{1}{2}\ln\left(2\pi e\,\sigma^2\right)}
$$

This is the well-known closed-form differential entropy of a Gaussian — notice it depends **only on $\sigma^2$**, not $\mu$ (translation invariance of entropy), and it grows logarithmically without bound as $\sigma^2\to\infty$ (a wider Gaussian is "more uncertain").

```python
import sympy as sp

x, mu, sigma, t = sp.symbols('x mu sigma t', real=True)
sigma_pos = sp.symbols('sigma', positive=True)

# --- Candidate MaxEnt density ---
p = (1/sp.sqrt(2*sp.pi*sigma_pos**2)) * sp.exp(-(x-mu)**2/(2*sigma_pos**2))

# 1) Normalization check: integral of p over R should be 1
norm_check = sp.integrate(p, (x, -sp.oo, sp.oo))
print("Normalization  ∫p dx        =", sp.simplify(norm_check))

# 2) Mean check: E[X] should equal mu
mean_check = sp.integrate(x*p, (x, -sp.oo, sp.oo))
print("Mean           E[X]         =", sp.simplify(mean_check))

# 3) Variance check: E[(X-mu)^2] should equal sigma^2
var_check = sp.integrate((x-mu)**2 * p, (x, -sp.oo, sp.oo))
print("Variance       E[(X-mu)^2]  =", sp.simplify(var_check))

# 4) Differential entropy: H[p] = -E[ln p]
lnp = sp.log(p)
H = sp.integrate(-p*lnp, (x, -sp.oo, sp.oo))
H_simplified = sp.simplify(H)
print("\nEntropy H[p]                =", H_simplified)

# Compare with closed form (1/2) ln(2*pi*e*sigma^2)
H_closed = sp.Rational(1,2)*sp.log(2*sp.pi*sp.E*sigma_pos**2)
print("Closed form (1/2)ln(2πeσ²)   =", H_closed)
print("Difference (should be 0)    :", sp.simplify(H_simplified - H_closed))

# --- Extra: verify Gaussian maximizes entropy numerically vs. other
#     distributions with the SAME mean & variance (e.g. Laplace, Uniform) ---
import numpy as np
from scipy import stats
from scipy.integrate import quad

mu_val, sigma_val = 0.0, 1.0

def entropy_numeric(pdf, a=-50, b=50):
    f = lambda x: -pdf(x)*np.log(pdf(x)) if pdf(x) > 0 else 0.0
    val, _ = quad(f, a, b, limit=200)
    return val

# Gaussian with mean mu_val, std sigma_val
gauss_pdf = lambda x: stats.norm.pdf(x, loc=mu_val, scale=sigma_val)

# Laplace matched to same variance: b = sigma/sqrt(2)
b_laplace = sigma_val/np.sqrt(2)
laplace_pdf = lambda x: stats.laplace.pdf(x, loc=mu_val, scale=b_laplace)

# Uniform matched to same variance: half-width a = sigma*sqrt(3)
a_unif = sigma_val*np.sqrt(3)
uniform_pdf = lambda x: stats.uniform.pdf(x, loc=mu_val-a_unif, scale=2*a_unif)

print("\nEntropy comparison (same mean=0, variance=1):")
print("Gaussian entropy :", entropy_numeric(gauss_pdf))
print("Laplace  entropy :", entropy_numeric(laplace_pdf))
print("Uniform  entropy :", entropy_numeric(uniform_pdf))
print("\n(Gaussian should have the LARGEST entropy among all three)")
```

# Q. Multiclass Logistic Regression: Automatic Classification of Emails

An e-commerce platform automatically routes inbound customer support emails to one of four mutually exclusive target destinations ($Y$):

* $Y = 0$ : **Finance**
* $Y = 1$ : **Technical**

Before final classification occurs, an upstream **AI Feature Extractor Agent** processes the raw text of each incoming email and outputs a discrete binary state observation vector $\mathbf{x} = (x_1, x_2, x_3)^T \in \mathscr{X}$, where:

* $x_1 = \mathbb{I}(\text{Financial context detected})$ (e.g., words like *"charge"*, *"invoice"*)
* $x_2 = \mathbb{I}(\text{Account access context detected})$ (e.g., words like *"password"*, *"login"*)
* $x_3 = \mathbb{I}(\text{Bug/malfunction context detected})$ (e.g., words like *"error"*, *"frozen"*)

---

Because human language inherently overlaps, the AI Agent's outputs are distributed non-deterministically across three common operational scenarios. The marginal probability of observing each state is fixed by historical logs to:

1. **Scenario A (Pure Financial):** $\mathbf{x}_A = (1, 0, 0)^T \implies P(\mathbf{x}_A) = 0.40$
2. **Scenario B (Pure Acces):** $\mathbf{x}_B = (0, 1, 0)^T \implies P(\mathbf{x}_B) = 0.25$
3. **Scenario C (Pure Technical):** $\mathbf{x}_C = (0, 0, 1)^T \implies P(\mathbf{x}_C) = 0.05$
3. **Scenario D (Access Technical):** $\mathbf{x}_D = (0, 1, 1)^T \implies P(\mathbf{x}_D) = 0.15$
3. **Scenario E (Billing Access):** $\mathbf{x}_E = (1, 1, 0)^T \implies P(\mathbf{x}_E) = 0.05$

---

Historical manual audits provide the following observed joint dataset expectations:

* Constraint 1: $\mathbb{E}[x_1 \cdot \mathbb{I}(Y=1)] = \mathbf{0.05}$
(Very low. If financial words are present, it is rarely routed to Technical support.)

* Constraint 2: $\mathbb{E}[x_2 \cdot \mathbb{I}(Y=1)] = \mathbf{0.15}$
(Moderate. Account access issues split down the middle depending on context.)

* Constraint 3: $\mathbb{E}[x_3 \cdot \mathbb{I}(Y=1)] = \mathbf{0.16}$
(High. If a bug or malfunction context is detected, it goes to the Tech team the vast majority of the time, utilizing $80\%$ of the $0.20$ max capacity).




# Answer

# Q. Multiclass Logistic Regression: Automatic Classification of Emails

## 1. Setup

**Note on the label space:** although the introduction mentions routing to "one of four" destinations, only two classes are actually defined in the problem: $Y=0$ (Finance) and $Y=1$ (Technical). This is therefore a **binary** MaxEnt/logistic classification problem (the direct binary specialization of the multiclass Softmax framework, exactly as in Section 4.5 of the notes).

The feature vector is $\mathbf{x} = (x_1, x_2, x_3)^T \in \{0,1\}^3$, with joint feature functions

$$
\mathbf{f}(\mathbf{x}, Y) = \mathbb{I}(Y=1)\,\mathbf{x}
$$

**Note on the marginal distribution $P(\mathbf{x})$:** the five given scenario probabilities sum to

$$
0.40+0.25+0.05+0.15+0.05 = 0.90 \neq 1
$$

Since $P(\mathbf{x})$ must be a valid probability distribution, we attribute the missing mass to the (unstated) "no-context-detected" scenario $\mathbf{x}_F = (0,0,0)^T$:

$$
P(\mathbf{x}_F) = 1 - 0.90 = 0.10
$$

This is the natural completion: an email with none of the three keyword contexts detected. Since $\mathbf{x}_F = \mathbf{0}$, it contributes **zero** to every constraint below, but it *is* needed for $P(\mathbf{x})$ to be a valid distribution and for $P(Y{=}1\mid\mathbf{x}_F)$ to be well defined.

| Scenario | $\mathbf{x}=(x_1,x_2,x_3)$ | $P(\mathbf{x})$ |
|---|---|---|
| A (Pure Financial) | $(1,0,0)$ | $0.40$ |
| B (Pure Access) | $(0,1,0)$ | $0.25$ |
| C (Pure Technical) | $(0,0,1)$ | $0.05$ |
| D (Access + Technical) | $(0,1,1)$ | $0.15$ |
| E (Billing + Access) | $(1,1,0)$ | $0.05$ |
| F (No context) | $(0,0,0)$ | $0.10$ |

## 2. The Conditional MaxEnt Lagrangian

Following the conditional-entropy MaxEnt derivation (Section 4.2–4.3 of the notes), we seek $P(Y\mid \mathbf{x})$ maximizing

$$
H(Y\mid X) = -\sum_{\mathbf{x}} \widetilde{P}(\mathbf{x}) \sum_{c=0}^{1} P(Y=c\mid \mathbf{x}) \ln P(Y=c\mid \mathbf{x})
$$

subject to $\sum_c P(Y=c\mid\mathbf{x}) = 1$ for every $\mathbf{x}$, and subject to matching the three empirical constraints:

$$
\mathbb{E}\big[x_1\,\mathbb{I}(Y=1)\big] = 0.05, \qquad
\mathbb{E}\big[x_2\,\mathbb{I}(Y=1)\big] = 0.15, \qquad
\mathbb{E}\big[x_3\,\mathbb{I}(Y=1)\big] = 0.16
$$

As derived generally in the notes, the stationary solution of this Lagrangian collapses (after eliminating the redundant weight vector via $\mathbf{w} = \mathbf{w}_1 - \mathbf{w}_0 \in \mathbb{R}^3$) to the **binary logistic (sigmoid) form**:

$$
P^*(Y=1\mid\mathbf{x}) = \frac{1}{1+e^{-\mathbf{w}\cdot\mathbf{x}}}, \qquad
P^*(Y=0\mid\mathbf{x}) = \frac{1}{1+e^{\mathbf{w}\cdot\mathbf{x}}}
$$

where $\mathbf{w}=(w_1,w_2,w_3)^T$ and $\mathbf{w}\cdot\mathbf{x} = w_1 x_1 + w_2 x_2 + w_3 x_3$ (no bias term, since none of the given constraints involve an intercept feature).

## 3. Explicit Constraint Equations

Writing $\sigma(z) = \dfrac{1}{1+e^{-z}}$ and substituting $\mathbf{w}\cdot\mathbf{x}$ for each scenario:

| Scenario | $\mathbf{w}\cdot\mathbf{x}$ |
|---|---|
| A $(1,0,0)$ | $w_1$ |
| B $(0,1,0)$ | $w_2$ |
| C $(0,0,1)$ | $w_3$ |
| D $(0,1,1)$ | $w_2+w_3$ |
| E $(1,1,0)$ | $w_1+w_2$ |
| F $(0,0,0)$ | $0$ |

The three constraint equations $\mathbb{E}[x_i\,\mathbb{I}(Y{=}1)] = \sum_{\mathbf{x}} P(\mathbf{x})\, x_i\, \sigma(\mathbf{w}\cdot\mathbf{x})$ become:

$$
\text{(1)} \quad 0.40\,\sigma(w_1) + 0.05\,\sigma(w_1+w_2) = 0.05
$$

$$
\text{(2)} \quad 0.25\,\sigma(w_2) + 0.15\,\sigma(w_2+w_3) + 0.05\,\sigma(w_1+w_2) = 0.15
$$

$$
\text{(3)} \quad 0.05\,\sigma(w_3) + 0.15\,\sigma(w_2+w_3) = 0.16
$$

This is a nonlinear system in $(w_1, w_2, w_3)$ with **no closed form**, so it is solved numerically.

*(Sanity check on constraint 3: the maximum possible value of $\mathbb{E}[x_3\mathbb{I}(Y{=}1)]$ occurs when $\sigma(w_3)=\sigma(w_2+w_3)=1$, giving $0.05+0.15=0.20$ — exactly the "max capacity" mentioned in the problem, and $0.16/0.20 = 80\%$, matching the stated description.)*

```python
import numpy as np
from scipy.optimize import fsolve

# Scenario probabilities
P_A, P_B, P_C, P_D, P_E = 0.40, 0.25, 0.05, 0.15, 0.05
P_F = 1 - (P_A + P_B + P_C + P_D + P_E)   # no-context scenario, = 0.10
print("P_F (no keyword context) =", P_F)

def sigmoid(z):
    return 1 / (1 + np.exp(-z))

def constraints(w):
    w1, w2, w3 = w
    sA = sigmoid(w1)        # x.w = w1
    sB = sigmoid(w2)        # x.w = w2
    sC = sigmoid(w3)        # x.w = w3
    sD = sigmoid(w2 + w3)   # x.w = w2 + w3
    sE = sigmoid(w1 + w2)   # x.w = w1 + w2

    eq1 = P_A*sA + P_E*sE            - 0.05   # E[x1 * I(Y=1)]
    eq2 = P_B*sB + P_D*sD + P_E*sE   - 0.15   # E[x2 * I(Y=1)]
    eq3 = P_C*sC + P_D*sD            - 0.16   # E[x3 * I(Y=1)]
    return [eq1, eq2, eq3]

w0_guess = [-2.0, 0.0, 1.5]
w_sol, info, ier, msg = fsolve(constraints, w0_guess, full_output=True, xtol=1e-14)

w1, w2, w3 = w_sol
print(f"w1 = {w1:.5f}")
print(f"w2 = {w2:.5f}")
print(f"w3 = {w3:.5f}")
print("Residuals of the 3 constraint equations:", constraints(w_sol))
print("Converged:", ier == 1, "-", msg)
```

```python
scenarios = {
    "A – Pure Financial (1,0,0)": (1, 0, 0),
    "B – Pure Access (0,1,0)":    (0, 1, 0),
    "C – Pure Technical (0,0,1)": (0, 0, 1),
    "D – Access+Technical (0,1,1)": (0, 1, 1),
    "E – Billing+Access (1,1,0)":   (1, 1, 0),
    "F – No context (0,0,0)":       (0, 0, 0),
}

print(f"{'Scenario':35s} {'w.x':>8s} {'P(Y=1|x)':>10s} {'P(Y=0|x)':>10s}")
for name, (x1, x2, x3) in scenarios.items():
    z = w1*x1 + w2*x2 + w3*x3
    p1 = sigmoid(z)
    print(f"{name:35s} {z:8.4f} {p1:10.4f} {1-p1:10.4f}")
```

## 4. Final MaxEnt Classifier

$$
\boxed{P^*(Y=1\mid \mathbf{x}) = \dfrac{1}{1+\exp\big(1.9725\,x_1 + 1.7762\,x_2 - 2.8804\,x_3\big)}}
$$

$$
P^*(Y=0\mid \mathbf{x}) = 1 - P^*(Y=1\mid \mathbf{x})
$$

## 5. Interpretation

- $w_1 \approx -1.97$ (large **negative**): detecting financial vocabulary strongly pushes the email **away** from Technical, confirming *"if financial words are present, it is rarely routed to Technical."*
- $w_3 \approx +2.88$ (large **positive**, and the largest-magnitude weight): detecting bug/malfunction vocabulary is the single strongest signal **toward** Technical routing, confirming the *"80% of the 0.20 max capacity"* behavior.
- $w_2 \approx -1.78$ (**negative**) is interesting: *access* keywords **alone** actually push toward Finance ($P(Y{=}1\mid \mathbf{x}_B) = 0.145$), but when access **and** technical keywords co-occur (Scenario D), the strongly positive $w_3$ dominates and flips the routing decisively toward Technical ($P(Y{=}1\mid \mathbf{x}_D) = 0.751$). This is exactly the *context-dependent* "split down the middle" behavior described for account-access emails — access alone reads as a billing/account issue, but access **combined with** a malfunction signal reads as a technical bug.
- For an email with **no detected keywords** (Scenario F), the model reverts to the fully unbiased $P(Y{=}1)=0.5$ — the correct MaxEnt behavior when there is no discriminating evidence at all.

# Q. Asset Pricing and Portfolio Optimization.



In standard finance, models often assume asset returns follow a perfect Normal (Gaussian) distribution. However, real-world markets experience extreme events, leading to "fat tails" and skewed returns. MaxEnt allows an analyst to infer the **least-biased probability distribution** of an asset's future price based *only* on what the market currently tells us, without forcing it into a rigid Gaussian shape.

**Background:**

You are a quantitative portfolio manager evaluating a highly volatile tech stock, **Asset X**, currently trading at $\$100$. You want to determine the risk-neutral probability distribution of its price, $S_T$, in exactly one month. This distribution will allow you to correctly price exotic derivatives and optimize your portfolio's risk exposure.

**Available Market Data (The Constraints):**

You do not know the true distribution, but you can observe the following macroeconomic aggregates and market-clearing prices from the options exchange:

1. **The Risk-Free Growth:** Based on the risk-free interest rate, the expected future price of the asset must match the forward price:

$$\mathbb{E}[S_T] = \$102$$


2. **The Market Volatility:** Based on standard historical variance, the expected squared price is constrained by:

$$\mathbb{E}[S_T^2] = 10,600$$


3. **Asymmetric Market Sentiment (The "Skew"):** Due to heavy trading on out-of-the-money put options, the market is pricing in a strong downside risk. The expected value of a specific logarithmic penalty (which captures downward crash risk) is constrained to:

$$\mathbb{E}[\ln(S_T)] = 4.60$$



**Objective:**

Formulate and solve the Maximum Entropy problem to find the continuous probability density function $p(S_T)$ for $S_T > 0$.



# Answer

# Q. Maximum Entropy Risk-Neutral Density for Asset X

## 1. Problem Setup

We want the least-biased continuous density $p(S_T)$ for $S_T \in (0,\infty)$, using a **uniform reference measure** $m(S_T)=1$ (no prior structural bias on the domain), subject to three macroscopic (market-implied) constraints:

$$
f_1(S_T) = S_T, \qquad f_2(S_T) = S_T^2, \qquad f_3(S_T) = \ln(S_T)
$$

$$
\mathbb{E}[f_1(S_T)] = \mathbb{E}[S_T] = 102, \qquad
\mathbb{E}[f_2(S_T)] = \mathbb{E}[S_T^2] = 10600, \qquad
\mathbb{E}[f_3(S_T)] = \mathbb{E}[\ln S_T] = 4.60
$$

together with the normalization constraint $\displaystyle\int_0^\infty p(S_T)\,dS_T = 1$.

## 2. Constructing the Functional Lagrangian

Following the continuous MaxEnt framework, we maximize the Shannon–Jaynes entropy

$$
H_c[p] = -\int_0^\infty p(S_T)\ln\!\left(\frac{p(S_T)}{m(S_T)}\right) dS_T ,\qquad m(S_T)=1
$$

subject to normalization and the three moment constraints. The Lagrangian functional is:

$$
\mathscr{L}[p] = -\int_0^\infty p(S_T)\ln p(S_T)\,dS_T
- \gamma\left(\int_0^\infty p(S_T)\,dS_T - 1\right)
- \lambda_1\left(\int_0^\infty p(S_T)\,S_T\,dS_T - 102\right)
- \lambda_2\left(\int_0^\infty p(S_T)\,S_T^2\,dS_T - 10600\right)
- \lambda_3\left(\int_0^\infty p(S_T)\,\ln S_T\,dS_T - 4.60\right)
$$

Taking the functional (Euler–Lagrange) derivative with respect to $p(S_T)$ and setting it to zero:

$$
\frac{\delta \mathscr{L}}{\delta p(S_T)} = -\ln p(S_T) - 1 - \gamma - \lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln S_T = 0
$$

Solving for $p(S_T)$:

$$
p^*(S_T) = \exp\!\big(-(1+\gamma)\big)\exp\!\left(-\lambda_1 S_T - \lambda_2 S_T^2 - \lambda_3 \ln S_T\right)
$$

## 3. The Maximum Entropy Density

Absorbing $e^{-(1+\gamma)}$ into a normalizing **partition function** $Z(\lambda_1,\lambda_2,\lambda_3)$, and noting $S_T^{-\lambda_3} = \exp(-\lambda_3 \ln S_T)$, the optimal density has the exponential-family form:

$$
\boxed{\;p^*(S_T) = \frac{1}{Z(\lambda_1,\lambda_2,\lambda_3)}\; S_T^{-\lambda_3}\, \exp\!\left(-\lambda_1 S_T - \lambda_2 S_T^2\right), \qquad S_T > 0\;}
$$

where the partition function is:

$$
Z(\lambda_1,\lambda_2,\lambda_3) = \int_0^\infty S_T^{-\lambda_3}\exp\!\left(-\lambda_1 S_T - \lambda_2 S_T^2\right) dS_T
$$

**Note on structure:** If $\lambda_3 = 0$, this collapses to a (possibly truncated) Gaussian — exactly the classic MaxEnt-Gaussian result from mean/variance constraints alone. The extra $S_T^{-\lambda_3}$ factor coming from the $\mathbb{E}[\ln S_T]$ constraint is precisely what **breaks the Gaussian symmetry** and injects the skew/fat-tail behavior the market is pricing through the put-skew.

As in the discrete/continuous theory, the multipliers are fixed by matching moments via the identity:

$$
-\frac{\partial \ln Z}{\partial \lambda_1} = \mathbb{E}[S_T] = 102, \qquad
-\frac{\partial \ln Z}{\partial \lambda_2} = \mathbb{E}[S_T^2] = 10600, \qquad
-\frac{\partial \ln Z}{\partial \lambda_3} = \mathbb{E}[\ln S_T] = 4.60
$$

These three nonlinear equations in $(\lambda_1,\lambda_2,\lambda_3)$ have **no closed-form solution** (unlike the pure-Gaussian case) because of the coupling between the linear, quadratic, and logarithmic terms in the exponent. We solve them **numerically**.

```python
import numpy as np
from scipy import integrate, optimize

# --- Market-implied constraints ---
target = np.array([102.0, 10600.0, 4.60])   # E[S], E[S^2], E[lnS]

# We substitute S = exp(t)  (t = ln S) to remove the S>0 boundary singularity
# and make the integrals numerically stable over the whole real line.
#
# p_unnorm(S) dS = S^{-l3} exp(-l1*S - l2*S^2) dS
#                = exp( (1-l3)*t - l1*e^t - l2*e^{2t} ) dt

def g(t, l1, l2, l3):
    return (1 - l3) * t - l1 * np.exp(t) - l2 * np.exp(2 * t)

def raw_moments(params):
    l1, l2, l3 = params
    Z, _     = integrate.quad(lambda t: np.exp(g(t, l1, l2, l3)), -60, 60, limit=800)
    m_S, _   = integrate.quad(lambda t: np.exp(t)   * np.exp(g(t, l1, l2, l3)), -60, 60, limit=800)
    m_S2, _  = integrate.quad(lambda t: np.exp(2*t) * np.exp(g(t, l1, l2, l3)), -60, 60, limit=800)
    m_lnS, _ = integrate.quad(lambda t: t           * np.exp(g(t, l1, l2, l3)), -60, 60, limit=800)
    return np.array([m_S, m_S2, m_lnS]) / Z, Z

def residual(params):
    m, _ = raw_moments(params)
    # scale residuals so all three constraints are weighted comparably
    return (m - target) / np.array([100.0, 10000.0, 4.6])

# initial guess ~ treating it as roughly Gaussian(mean=102, var=196)
guess = np.array([-0.5, 0.0025, 0.0])

sol = optimize.least_squares(residual, guess, method='lm',
                              xtol=1e-15, ftol=1e-15, gtol=1e-15, max_nfev=10000)

lambda1, lambda2, lambda3 = sol.x
moments_check, Z = raw_moments(sol.x)

print(f"lambda_1 = {lambda1:.6f}")
print(f"lambda_2 = {lambda2:.6f}")
print(f"lambda_3 = {lambda3:.6f}")
print(f"Z        = {Z:.6e}")
print(f"Check E[S], E[S^2], E[lnS] = {moments_check}")
```

## 4. Final Risk-Neutral MaxEnt Density

Substituting the solved multipliers back into the general form gives the explicit density for Asset X's price in one month:

$$
p^*(S_T) \;=\; \frac{1}{3.294\times10^{11}}\; S_T^{-1.2832}\;
\exp\!\left(0.5538\,S_T - 0.002651\,S_T^2\right), \qquad S_T>0
$$

This is **not** a simple Gaussian — it belongs to a generalized exponential-power family (a mixture of the "Gamma-type" behavior from the $S_T^{-\lambda_3}$ term and the "Gaussian-type" decay from the $S_T^2$ term). It is the continuous analogue of a **Generalized Inverse Gaussian / generalized Gamma-Gaussian hybrid**, which arises naturally whenever a mean, a variance, *and* a log-moment are all constrained simultaneously.

```python
import numpy as np
from scipy import integrate
import matplotlib.pyplot as plt

l1, l2, l3 = lambda1, lambda2, lambda3

def pdf_S(S):
    return np.exp(-l3*np.log(S) - l1*S - l2*S**2) / Z

# Central moments / skewness (via t = ln S substitution for stability)
def moment_power(k):
    val, _ = integrate.quad(lambda t: np.exp(k*t) * np.exp(g(t, l1, l2, l3)), -60, 60, limit=800)
    return val / Z

mean = moment_power(1)
m2   = moment_power(2)
m3   = moment_power(3)
var  = m2 - mean**2
sd   = np.sqrt(var)
skew = (m3 - 3*mean*m2 + 2*mean**3) / sd**3

print(f"Mean       = {mean:.4f}   (target 102)")
print(f"Std Dev    = {sd:.4f}     (implied by Var = E[S^2]-E[S]^2 = {var:.2f})")
print(f"Skewness   = {skew:.4f}   <-- negative => left (downside) fat tail, as market implies")

# --- Plot: MaxEnt density vs. naive Gaussian(102, 14^2) ---
S = np.linspace(30, 200, 2000)
p_maxent = pdf_S(S)

from scipy.stats import norm
p_gauss = norm.pdf(S, loc=102, scale=14)

plt.figure(figsize=(9,5))
plt.plot(S, p_maxent, label="MaxEnt density $p^*(S_T)$ (mean, var, and $\\ln S$ matched)", lw=2.5)
plt.plot(S, p_gauss, '--', label="Naive Gaussian(102, 14²) (mean & var only)", lw=2)
plt.axvline(100, color='gray', ls=':', label="Current price $S_0=100$")
plt.xlabel("$S_T$ (price in 1 month)")
plt.ylabel("Density")
plt.title("MaxEnt Risk-Neutral Density vs. Gaussian Assumption")
plt.legend()
plt.grid(alpha=0.3)
plt.show()
```

## 5. Interpretation for Portfolio Management

- **Mean and variance match exactly** ($\mathbb{E}[S_T]=102$, $\text{Var}(S_T)=196 \Rightarrow \sigma = 14$), same as a Gaussian model would give.
- **But the skewness is negative** (≈ $-0.10$), *not zero as a Gaussian would force*. This comes entirely from the third constraint, $\mathbb{E}[\ln S_T]=4.60$, which encodes the market's downside (put-skew) pricing information.
- Practically: the MaxEnt density has a **longer/heavier left tail** than the Gaussian with the same mean and variance — i.e., it assigns more probability mass to a crash scenario, consistent with heavy OTM put buying.
- This $p^*(S_T)$ is the "least-biased" (max-entropy) distribution consistent with *exactly* the market data given — no more, no less. It should be used (instead of assuming Normality) to price other derivatives on Asset X and to compute Value-at-Risk / Expected Shortfall for the portfolio, since it correctly captures the fat-tail/skew risk that a Gaussian model would understate.
