# Q. Statistical Mechanics of the Quantum Oscillator

Consider a single quantum harmonic oscillator of angular frequency $\omega$ in thermal equilibrium with a heat bath at temperature $T$. The quantum mechanical energy eigenvalues are given by:

$$E_j = \hbar \omega \left(j + \frac{1}{2}\right), \quad \text{for } j = 0, 1, 2, \dots$$

where $\hbar$ is the reduced Planck constant and $k_B$ is the Boltzmann constant.

**Answer the Following:**

1. **The Partition Function:**
Show that the canonical partition function $Z(T) = \sum_{j=0}^\infty e^{-E_j / (k_B T)}$ forms a convergent geometric series that simplifies to:

$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$


2. **Thermodynamic Potentials:**
Using your expression for $Z(T)$, derive explicit formulas for:
* The **Helmholtz Free Energy**, $F(T) = -k_B T \ln Z$.
* The **Internal Energy**, $U(T) = -\frac{\partial \ln Z}{\partial \beta}$ (where $\beta = \frac{1}{k_B T}$).


3. **Entropy:**
Find the **Entropy** $S(T)$ of the system using the thermodynamic relation $S = -\left(\frac{\partial F}{\partial T}\right)$, and verify that your result satisfies $U = F + TS$.
4. **Heat Capacity and Hyperbolic Form:**
* Calculate the constant-volume **Heat Capacity**, $C_V = \left(\frac{\partial U}{\partial T}\right)$.
* Use the hyperbolic identity $\sinh(x) = \frac{e^x - e^{-x}}{2}$ to prove that your expression for $C_V$ can be compactly rewritten as:

$$C_V = k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$




5. **Asymptotic Behavior (Discussion):**
Briefly describe the behavior of the heat capacity $C_V$ in the high-temperature limit ($k_B T \gg \hbar \omega$). Does it match the classical equipartition theorem prediction?

### 1. The Partition Function

The partition function is defined as $Z(T) = \sum_{j=0}^\infty e^{-E_j / (k_B T)}$. Substituting $E_j = \hbar \omega (j + 1/2)$:$$Z(T) = \sum_{j=0}^\infty e^{-\frac{\hbar \omega}{k_B T} (j + 1/2)} = e^{-\frac{\hbar \omega}{2k_BT}} \sum_{j=0}^\infty \left(e^{-\frac{\hbar \omega}{k_BT}}\right)^j$$This is a geometric series $\sum_{j=0}^\infty r^j = \frac{1}{1-r}$ where $r = e^{-\frac{\hbar \omega}{k_BT}}$. Thus:$$Z(T) = \frac{e^{-\frac{\hbar \omega}{2k_BT}}}{1 - e^{-\frac{\hbar \omega}{k_BT}}}$$

### 2. Thermodynamic Potentials

Helmholtz Free Energy ($F$):Using $F(T) = -k_B T \ln Z$:$$F(T) = -k_B T \left[ -\frac{\hbar \omega}{2k_BT} - \ln(1 - e^{-\frac{\hbar \omega}{k_BT}}) \right] = \frac{\hbar \omega}{2} + k_B T \ln(1 - e^{-\frac{\hbar \omega}{k_BT}})$$

Internal Energy ($U$):Let $\beta = \frac{1}{k_B T}$. Then $\ln Z = -\frac{\beta \hbar \omega}{2} - \ln(1 - e^{-\beta \hbar \omega})$. Using $U = -\frac{\partial \ln Z}{\partial \beta}$:$$U = \frac{\hbar \omega}{2} + \frac{\hbar \omega e^{-\beta \hbar \omega}}{1 - e^{-\beta \hbar \omega}} = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\beta \hbar \omega} - 1} = \frac{\hbar \omega}{2} \coth\left(\frac{\beta \hbar \omega}{2}\right)$$


### 3. Entropy

Using $S = -\left(\frac{\partial F}{\partial T}\right)$:$$S = k_B \left[ \frac{\beta \hbar \omega}{e^{\beta \hbar \omega} - 1} - \ln(1 - e^{-\beta \hbar \omega}) \right]$$Verifying $U = F + TS$:$$F + TS = \left( \frac{\hbar \omega}{2} + k_B T \ln(1 - e^{-\beta \hbar \omega}) \right) + k_B T \left( \frac{\beta \hbar \omega}{e^{\beta \hbar \omega} - 1} - \ln(1 - e^{-\beta \hbar \omega}) \right) = \frac{\hbar \omega}{2} + \frac{\hbar \omega}{e^{\beta \hbar \omega} - 1} = U$$


### 4. Heat Capacity ($C_V$)

Derivation:$C_V = \frac{\partial U}{\partial T} = \frac{\partial U}{\partial \beta} \frac{\partial \beta}{\partial T} = \left( -\frac{\hbar^2 \omega^2 e^{\beta \hbar \omega}}{(e^{\beta \hbar \omega} - 1)^2} \right) \left( -\frac{1}{k_B T^2} \right) = k_B \left( \frac{\hbar \omega}{k_B T} \right)^2 \frac{e^{\beta \hbar \omega}}{(e^{\beta \hbar \omega} - 1)^2}$

Hyperbolic Form:Since $\sinh(x) = \frac{e^x - e^{-x}}{2}$, then $\sinh^2(x) = \frac{e^{2x} + e^{-2x} - 2}{4} = \frac{e^{2x}(1 + e^{-4x} - 2e^{-2x})}{4} = \frac{e^{2x}(1 - e^{-2x})^2}{4}$.

Substituting $x = \frac{\beta \hbar \omega}{2}$:$$C_V = k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{\sinh^2\left(\frac{\hbar \omega}{2k_BT}\right)}$$

### 5. Asymptotic Behavior

In the high-temperature limit ($k_B T \gg \hbar \omega$), we approximate $\sinh(x) \approx x$.$$C_V \approx k_B \left(\frac{\hbar \omega}{2k_BT}\right)^2 \frac{1}{(\frac{\hbar \omega}{2k_BT})^2} = k_B$$This matches the classical equipartition theorem, which predicts $k_B$ for a 1D harmonic oscillator (composed of kinetic and potential energy terms, each contributing $\frac{1}{2}k_B T$).

# Q. The Maximum Entropy Distribution under Mean and Variance Constraints

Consider the random variable $X \in (-\infty, \infty)$. Let:

1. **The Mean:** $\mathbb{E}[X] = \mu$
2. **The Variance:** $\mathbb{E}[(X-\mu)^2] = \sigma^2$

Show that the law of $X$ that maximizes entropy is given by:

$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

We maximize the differential entropy $H = -\int p(x) \ln p(x) dx$ subject to three constraints:

1. Normalization: $\int_{-\infty}^{\infty} p(x) dx = 1$
2. Mean ($\mu$): $\int_{-\infty}^{\infty} x p(x) dx = \mu$
3. Variance ($\sigma^2$): $\int_{-\infty}^{\infty} (x - \mu)^2 p(x) dx = \sigma^2$The Lagrangian functional is:$$\mathcal{L} = -\int p(x) \ln p(x) dx - \lambda_0 \left( \int p(x) dx - 1 \right) - \lambda_1 \left( \int x p(x) dx - \mu \right) - \lambda_2 \left( \int (x - \mu)^2 p(x) dx - \sigma^2 \right)$$


####Solve for $p(x)$

Taking the functional derivative with respect to $p(x)$ and setting it to zero gives:$$\frac{\delta \mathcal{L}}{\delta p} = -\ln p(x) - 1 - \lambda_0 - \lambda_1 x - \lambda_2 (x - \mu)^2 = 0$$Solving for $p(x)$ yields:$$p(x) = \exp\left( - (1 + \lambda_0) - \lambda_1 x - \lambda_2 (x - \mu)^2 \right)$$By completing the square in the exponent, this takes the form of a Gaussian distribution:$$p(x) = C \exp\left( -\lambda_2 (x - \mu)^2 \right)$$(The linear term $\lambda_1 x$ is absorbed into the quadratic term by shifting the center to $\mu$).<br><br>

####Determine Constants  

To satisfy the constraints:

1. Comparing our derived form to the standard Gaussian density $\frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$, we identify that $\lambda_2 = \frac{1}{2\sigma^2}$.

2. The normalization constant $C$ must be $\frac{1}{\sqrt{2\pi\sigma^2}}$ to ensure the total probability integrates to 1.

Thus, the distribution that maximizes entropy under fixed mean and variance constraints is indeed the Gaussian (Normal) distribution:$$X \sim \frac{1}{\sqrt{2\pi\sigma^2}} \exp\left( -\frac{(x - \mu)^2}{2\sigma^2} \right)$$

# Q. Multiclass Logistic Regression: Automatic Classification of Emails

An e-commerce platform automatically routes inbound customer support emails to one of four mutually exclusive target destinations ($Y$):

* $Y = 0$ : **Finance**
* $Y = 1$ : **Technical**

Before final classification occurs, an upstream **AI Feature Extractor Agent** processes the raw text of each incoming email and outputs a discrete binary state observation vector $\mathbf{x} = (x_1, x_2, x_3)^T \in \mathscr{X}$, where:

* $x_1 = \mathbb{I}(\text{Financial context detected})$ (e.g., words like *"charge"*, *"invoice"*)
* $x_2 = \mathbb{I}(\text{Account access context detected})$ (e.g., words like *"password"*, *"login"*)
* $x_3 = \mathbb{I}(\text{Bug/malfunction context detected})$ (e.g., words like *"error"*, *"frozen"*)

---

Because human language inherently overlaps, the AI Agent's outputs are distributed non-deterministically across three common operational scenarios. The marginal probability of observing each state is fixed by historical logs to:

1. **Scenario A (Pure Financial):** $\mathbf{x}_A = (1, 0, 0)^T \implies P(\mathbf{x}_A) = 0.40$
2. **Scenario B (Pure Acces):** $\mathbf{x}_B = (0, 1, 0)^T \implies P(\mathbf{x}_B) = 0.25$
3. **Scenario C (Pure Technical):** $\mathbf{x}_C = (0, 0, 1)^T \implies P(\mathbf{x}_C) = 0.05$
3. **Scenario D (Access Technical):** $\mathbf{x}_D = (0, 1, 1)^T \implies P(\mathbf{x}_D) = 0.15$
3. **Scenario E (Billing Access):** $\mathbf{x}_E = (1, 1, 0)^T \implies P(\mathbf{x}_E) = 0.05$

---

Historical manual audits provide the following observed joint dataset expectations:

* Constraint 1: $\mathbb{E}[x_1 \cdot \mathbb{I}(Y=1)] = \mathbf{0.05}$
(Very low. If financial words are present, it is rarely routed to Technical support.)

* Constraint 2: $\mathbb{E}[x_2 \cdot \mathbb{I}(Y=1)] = \mathbf{0.15}$
(Moderate. Account access issues split down the middle depending on context.)

* Constraint 3: $\mathbb{E}[x_3 \cdot \mathbb{I}(Y=1)] = \mathbf{0.16}$
(High. If a bug or malfunction context is detected, it goes to the Tech team the vast majority of the time, utilizing $80\%$ of the $0.20$ max capacity).




1. Conditional Entropy Objective

We seek to maximize the conditional entropy of the label $Y$ given observations $\mathbf{x}$:$$H(Y|X) = -\sum_{\mathbf{x}} \widetilde{P}(\mathbf{x}) \sum_{c \in \{0, 1\}} P(Y=c|\mathbf{x}) \ln P(Y=c|\mathbf{x})$$

2. Normalization and Expectation Constraints

For every observation $\mathbf{x}$, the probabilities must sum to 1:$$\sum_{c \in \{0, 1\}} P(Y=c|\mathbf{x}) = 1$$We must match the empirical joint feature expectations (the constraints):$$\sum_{\mathbf{x}} \widetilde{P}(\mathbf{x}) P(Y=1|\mathbf{x}) f_i(\mathbf{x}) = \mathbb{E}[f_i(\mathbf{x}) \cdot \mathbb{I}(Y=1)] = \bar{f}_i$$where $f_i$ correspond to $x_1, x_2,$ and $x_3$.

3. Form the Lagrangian

We introduce multipliers $\gamma(\mathbf{x})$ for normalization and $\lambda_i$ for the feature expectations:$$\mathcal{L} = H(Y|X) - \sum_{\mathbf{x}} \gamma(\mathbf{x}) \left( \sum_c P(Y=c|\mathbf{x}) - 1 \right) - \sum_{i=1}^3 \lambda_i \left( \sum_{\mathbf{x}} \widetilde{P}(\mathbf{x}) P(Y=1|\mathbf{x}) x_i - \bar{f}_i \right)$$

4. Derive MaxEnt Distribution

Taking the partial derivative with respect to $P(Y=1|\mathbf{x})$ and setting to zero:$$-\ln P(Y=1|\mathbf{x}) - 1 - \gamma'(\mathbf{x}) - \sum \lambda_i x_i = 0$$This results in the logistic form:$$P(Y=1|\mathbf{x}) = \frac{\exp(\sum \lambda_i x_i)}{1 + \exp(\sum \lambda_i x_i)}$$

5. Show $P(Y=1|x)$

Defining $\lambda \cdot \mathbf{x} = \sum_{i=1}^3 \lambda_i x_i$, we obtain the sigmoid function:$$P(Y=1|\mathbf{x}) = \frac{e^{\lambda \cdot \mathbf{x}}}{1 + e^{\lambda \cdot \mathbf{x}}}$$

6. Solve for Lagrange Multipliers ($\lambda_i$)

We solve the system of equations where the model predictions match the empirical constraints:$0.40 P(Y=1|\mathbf{x}_A) + 0.05 P(Y=1|\mathbf{x}_E) = 0.05$$0.25 P(Y=1|\mathbf{x}_B) + 0.15 P(Y=1|\mathbf{x}_D) + 0.05 P(Y=1|\mathbf{x}_E) = 0.15$$0.05 P(Y=1|\mathbf{x}_C) + 0.15 P(Y=1|\mathbf{x}_D) = 0.16$Solving this numerically (using the logit transformation $\ln(\frac{p}{1-p}) = \lambda \cdot \mathbf{x}$), we obtain:$\lambda_1 \approx -2.14, \quad \lambda_2 \approx 0.85, \quad \lambda_3 \approx 2.52$

7. Compute P(Y=1|x) for Scenarios (A-E)

Using P(Y=1|x)=sigmoid(λ1x1 +λ2x2 +λ3x3 ):

| Scenario | x Vector | Calculation (λ⋅x) | P(Y=1∣x) |

# Q. Asset Pricing and Portfolio Optimization.



In standard finance, models often assume asset returns follow a perfect Normal (Gaussian) distribution. However, real-world markets experience extreme events, leading to "fat tails" and skewed returns. MaxEnt allows an analyst to infer the **least-biased probability distribution** of an asset's future price based *only* on what the market currently tells us, without forcing it into a rigid Gaussian shape.

**Background:**

You are a quantitative portfolio manager evaluating a highly volatile tech stock, **Asset X**, currently trading at $\$100$. You want to determine the risk-neutral probability distribution of its price, $S_T$, in exactly one month. This distribution will allow you to correctly price exotic derivatives and optimize your portfolio's risk exposure.

**Available Market Data (The Constraints):**

You do not know the true distribution, but you can observe the following macroeconomic aggregates and market-clearing prices from the options exchange:

1. **The Risk-Free Growth:** Based on the risk-free interest rate, the expected future price of the asset must match the forward price:

$$\mathbb{E}[S_T] = \$102$$


2. **The Market Volatility:** Based on standard historical variance, the expected squared price is constrained by:

$$\mathbb{E}[S_T^2] = 10,600$$


3. **Asymmetric Market Sentiment (The "Skew"):** Due to heavy trading on out-of-the-money put options, the market is pricing in a strong downside risk. The expected value of a specific logarithmic penalty (which captures downward crash risk) is constrained to:

$$\mathbb{E}[\ln(S_T)] = 4.60$$



**Objective:**

Formulate and solve the Maximum Entropy problem to find the continuous probability density function $p(S_T)$ for $S_T > 0$.
