# Q. Analytical Derivation

Consider the filter model
\begin{align*}
x^{-}_k &= A_{k-1}\,x^{+}_{k-1} + G_{k-1}\,w_{k-1}, \\
y^{-}_k &= H_k\,x^{-}_k + z_k,
\end{align*}
where
$$x^{+}_{k-1} \sim \mathscr{N}(m_{k-1}, P_{k-1}),$$

Answer the followings:

1. Show that
$$x_k^- \sim \mathscr{N}(m_k^-, P_k^-),$$
where
$$m_k^- = A_{k-1}m_{k-1},$$
and
$$
P_k^- = A_{k-1}P_{k-1}A_{k-1}^T + G_{k-1}\Sigma_p G_{k-1}^T.
$$

2. Show that
$$y_k^- \sim \mathscr{N}(H_k m_k^-, H_k P_k^- H_k^T + \Sigma_m).$$

3. Show that
$$\begin{bmatrix} x_k^- \\ y^{-}_k \end{bmatrix} \sim \mathscr{N}\left(
\begin{bmatrix} m_k^- \\ H_k m_k^- \end{bmatrix},
\begin{bmatrix} P_k^- & P_k^- H_k^T \\ H_k P_k^- & H_k P_k^- H_k^T + \Sigma_m \end{bmatrix}
\right).$$

4. Show that
$$x^{+}_k\triangleq (x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}) \sim \mathscr{N}(m_k, P_k),$$
where the updated mean $m_k$ and updated covariance $P_k$ are given by:
\begin{align*}
K_k &\triangleq P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}, \\
m_k &= m_k^- + K_k (y^{\mathrm{obs}}_{k} - H_k m_k^-), \\
P_k &= (I - K_k H_k) P_k^-.
\end{align*}

5. Find
$\mathbb{E}[x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k}]$  and
$\text{Var}(x^{-}_k \mid y^{-}_k = y^{\mathrm{obs}}_{k})$.

1. Prediction Step (Time Update)Given $x_k^- = A_{k-1}x_{k-1}^+ + G_{k-1}w_{k-1}$ and $x_{k-1}^+ \sim \mathscr{N}(m_{k-1}, P_{k-1})$, with $w_{k-1} \sim \mathscr{N}(0, \Sigma_p)$:Mean:$$m_k^- \triangleq \mathbb{E}[x_k^-] = \mathbb{E}[A_{k-1}x_{k-1}^+ + G_{k-1}w_{k-1}] = A_{k-1}m_{k-1} + G_{k-1}\mathbb{E}[w_{k-1}] = A_{k-1}m_{k-1}$$Covariance:$$P_k^- \triangleq \text{Var}(x_k^-) = \text{Var}(A_{k-1}x_{k-1}^+ + G_{k-1}w_{k-1}) = A_{k-1}\text{Var}(x_{k-1}^+)A_{k-1}^T + G_{k-1}\text{Var}(w_{k-1})G_{k-1}^T$$$$P_k^- = A_{k-1}P_{k-1}A_{k-1}^T + G_{k-1}\Sigma_p G_{k-1}^T$$

2. Measurement PredictionGiven $y_k^- = H_k x_k^- + z_k$ and $x_k^- \sim \mathscr{N}(m_k^-, P_k^-)$, with $z_k \sim \mathscr{N}(0, \Sigma_m)$:Mean:$$\mathbb{E}[y_k^-] = H_k \mathbb{E}[x_k^-] + \mathbb{E}[z_k] = H_k m_k^-$$Variance:$$\text{Var}(y_k^-) = \text{Var}(H_k x_k^- + z_k) = H_k \text{Var}(x_k^-) H_k^T + \text{Var}(z_k) = H_k P_k^- H_k^T + \Sigma_m$$

3. Joint Prior DistributionThe joint distribution of the predicted state $x_k^-$ and the measurement $y_k^-$ is Gaussian. We need the cross-covariance:$$\text{Cov}(x_k^-, y_k^-) = \text{Cov}(x_k^-, H_k x_k^- + z_k) = \text{Cov}(x_k^-, x_k^-) H_k^T = P_k^- H_k^T$$Assembling the block matrix:$$\begin{bmatrix} x_k^- \\ y^{-}_k \end{bmatrix} \sim \mathscr{N}\left( \begin{bmatrix} m_k^- \\ H_k m_k^- \end{bmatrix}, \begin{bmatrix} P_k^- & P_k^- H_k^T \\ H_k P_k^- & H_k P_k^- H_k^T + \Sigma_m \end{bmatrix} \right)$$

4. Measurement Update (Correction Step)Using the conditional properties of a joint multivariate Gaussian:If $\begin{bmatrix} X \\ Y \end{bmatrix} \sim \mathscr{N}\left( \begin{bmatrix} \mu_X \\ \mu_Y \end{bmatrix}, \begin{bmatrix} \Sigma_{XX} & \Sigma_{XY} \\ \Sigma_{YX} & \Sigma_{YY} \end{bmatrix} \right)$, then the conditional distribution $(X|Y=y)$ is $\mathscr{N}(\mu, \Sigma)$ where:$$\mu = \mu_X + \Sigma_{XY}\Sigma_{YY}^{-1}(y - \mu_Y)$$$$\Sigma = \Sigma_{XX} - \Sigma_{XY}\Sigma_{YY}^{-1}\Sigma_{YX}$$Substituting $X=x_k^-$, $Y=y_k^-$, $y=y_k^{\text{obs}}$, $\mu_X = m_k^-$, $\mu_Y = H_k m_k^-$, $\Sigma_{XY} = P_k^- H_k^T$, and $\Sigma_{YY} = H_k P_k^- H_k^T + \Sigma_m$:Kalman Gain:$$K_k = \Sigma_{XY}\Sigma_{YY}^{-1} = P_k^- H_k^T (H_k P_k^- H_k^T + \Sigma_m)^{-1}$$Updated Mean:$$m_k = m_k^- + K_k(y_k^{\text{obs}} - H_k m_k^-)$$Updated Covariance:$$P_k = P_k^- - K_k H_k P_k^- = (I - K_k H_k)P_k^-$$

5. Expected Value and VarianceBased on the derivations in Part4:Conditional Mean:$$\mathbb{E}[x_k^- \mid y_k^- = y_k^{\text{obs}}] = m_k = m_k^- + K_k(y_k^{\text{obs}} - H_k m_k^-)$$Conditional Variance:$$\text{Var}(x_k^- \mid y_k^- = y_k^{\text{obs}}) = P_k = (I - K_k H_k)P_k^-$$

```python
import numpy as np


def kalman_filter_step(mk_minus, Pk_minus, y_obs, Ak, Gk, Hk, Sigma_p, Sigma_m, wk_mean=0):
    """
    Performs a full Kalman Filter iteration (Prediction + Update).
    """
    # 1. Prediction Step
    mk_pred = Ak @ mk_minus
    Pk_pred = Ak @ Pk_minus @ Ak.T + Gk @ Sigma_p @ Gk.T

    # 2. Update Step
    # Innovation Covariance S = H * P_pred * H^T + R
    Sk = Hk @ Pk_pred @ Hk.T + Sigma_m

    # Kalman Gain K = P_pred * H^T * inv(S)
    Kk = Pk_pred @ Hk.T @ np.linalg.inv(Sk)

    # Innovation (residual) v = y_obs - H * mk_pred
    vk = y_obs - (Hk @ mk_pred)

    # Updated state and covariance
    mk_upd = mk_pred + Kk @ vk
    Pk_upd = (np.eye(len(mk_pred)) - Kk @ Hk) @ Pk_pred

    return mk_upd, Pk_upd

# 1. Define system dimensions and sampling time
dt = 0.1

# 2. Define System Matrices (A, H, G, Q, R)
A = np.array([[1, 0, dt, 0],
              [0, 1, 0, dt],
              [0, 0, 1, 0],
              [0, 0, 0, 1]])

H = np.array([[1, 0, 0, 0],
              [0, 1, 0, 0]])

G = np.array([[0.5*dt**2, 0],
              [0, 0.5*dt**2],
              [dt, 0],
              [0, dt]])

Q = 1e-4 * np.eye(2) # Process noise covariance Sigma_p
R = 0.5**2 * np.eye(2) # Measurement noise covariance Sigma_m

# 3. Define Initial State and Covariance
m_prev = np.array([0, 0, 0, 0]) # Initial mean
P_prev = np.eye(4) * 100        # Initial covariance

# 4. Define Observation (Sample measurement)
y_obs = np.array([1.0, 1.0])

mk, Pk = kalman_filter_step(m_prev, P_prev, y_obs, A, G, H, Q, R)

print("Updated Mean:", mk)
print("Updated Covariance:", Pk)
```

# Q. 2D-Position Estimation

## Part A

Consider a two-dimensional constant-velocity dynamical system. The hidden state at time step $k$ is

$$
x_k =
\begin{bmatrix}
p_x(k)\\
p_y(k)\\
v_x(k)\\
v_y(k)
\end{bmatrix},
$$

where $(p_x(k),p_y(k))$ denote the position components and $(v_x(k),v_y(k))$ denote the velocity components.

The measurement consists only of the two position components:

$$
y_k =
\begin{bmatrix}
p_x^{\mathrm{meas}}(k)\\
p_y^{\mathrm{meas}}(k)
\end{bmatrix}.
$$

The linear Gaussian state-space filter model is

$$
x^-_k = A x^+_{k-1} + G w_{k-1},
$$

$$
y^-_k = Hx^+_k + z_k,
$$

where

$$
w_{k-1} \sim \mathscr{N}(0,\Sigma_p),
\qquad
z_k \sim \mathscr{N}(0,\Sigma_m).
$$

The process noise sequence $w_k$, measurement noise sequence $z_k$, and the initial state are assumed mutually independent.

Show that:

$$
A =
\begin{bmatrix}
1 & 0 & \Delta t & 0\\
0 & 1 & 0 & \Delta t\\
0 & 0 & 1 & 0\\
0 & 0 & 0 & 1
\end{bmatrix},
$$



$$
H =
\begin{bmatrix}
1 & 0 & 0 & 0\\
0 & 1 & 0 & 0
\end{bmatrix},
$$

and

$$
G =
\begin{bmatrix}
\frac{1}{2}\Delta t^2 & 0\\
0 & \frac{1}{2}\Delta t^2\\
\Delta t & 0\\
0 & \Delta t
\end{bmatrix}.
$$

To show these matrices for the 2D constant-velocity (CV) system, we analyze the kinematic equations over a sampling interval $\Delta t$.

1. Derivation of State Transition Matrix $A$The kinematic equations for constant velocity are:$p_x(k) = p_x(k-1) + \Delta t \, v_x(k-1)$$p_y(k) = p_y(k-1) + \Delta t \, v_y(k-1)$$v_x(k) = v_x(k-1)$$v_y(k) = v_y(k-1)$Writing these as a system of linear equations $x_k = A x_{k-1}$ (ignoring noise):$$\begin{bmatrix} p_x(k) \\ p_y(k) \\ v_x(k) \\ v_y(k) \end{bmatrix} =
\begin{bmatrix} 1 & 0 & \Delta t & 0 \\ 0 & 1 & 0 & \Delta t \\ 0 & 0 & 1 & 0 \\ 0 & 0 & 0 & 1 \end{bmatrix}
\begin{bmatrix} p_x(k-1) \\ p_y(k-1) \\ v_x(k-1) \\ v_y(k-1) \end{bmatrix}$$Thus, the state transition matrix $A$ is:$$A = \begin{bmatrix} 1 & 0 & \Delta t & 0 \\ 0 & 1 & 0 & \Delta t \\ 0 & 0 & 1 & 0 \\ 0 & 0 & 0 & 1 \end{bmatrix}$$

2. Derivation of Measurement Matrix $H$The measurement $y_k$ contains only the position components:$$y_k = \begin{bmatrix} p_x(k) \\ p_y(k) \end{bmatrix} = \begin{bmatrix} 1 & 0 & 0 & 0 \\ 0 & 1 & 0 & 0 \end{bmatrix} \begin{bmatrix} p_x(k) \\ p_y(k) \\ v_x(k) \\ v_y(k) \end{bmatrix}$$This defines the observation matrix $H$:$$H = \begin{bmatrix} 1 & 0 & 0 & 0 \\ 0 & 1 & 0 & 0 \end{bmatrix}$$

3. Derivation of Noise Input Matrix $G$Assuming random acceleration $w_{k-1} = [a_x(k-1), a_y(k-1)]^T$, the equations of motion over $\Delta t$ include the displacement due to acceleration:$p_x(k) = p_x(k-1) + \Delta t \, v_x(k-1) + \frac{1}{2}\Delta t^2 a_x(k-1)$$p_y(k) = p_y(k-1) + \Delta t \, v_y(k-1) + \frac{1}{2}\Delta t^2 a_y(k-1)$$v_x(k) = v_x(k-1) + \Delta t \, a_x(k-1)$$v_y(k) = v_y(k-1) + \Delta t \, a_y(k-1)$Representing the noise contribution $G w_{k-1}$:$$\begin{bmatrix} p_x(k) \\ p_y(k) \\ v_x(k) \\ v_y(k) \end{bmatrix} = A x_{k-1} +
\begin{bmatrix} \frac{1}{2}\Delta t^2 & 0 \\ 0 & \frac{1}{2}\Delta t^2 \\ \Delta t & 0 \\ 0 & \Delta t \end{bmatrix}
\begin{bmatrix} a_x(k-1) \\ a_y(k-1) \end{bmatrix}$$Thus, the matrix $G$ is:$$G = \begin{bmatrix} \frac{1}{2}\Delta t^2 & 0 \\ 0 & \frac{1}{2}\Delta t^2 \\ \Delta t & 0 \\ 0 & \Delta t \end{bmatrix}$$

```python
import numpy as np

def filter_gps_measurements(Y_obs, dt=0.1, q=1e-4, r=0.5):
    """
    Filters 2D GPS position measurements using a Constant Velocity Kalman Filter.

    Y_obs: np.ndarray of shape (T, 2) - [px, py] measurements
    dt: float - sampling period
    q: float - acceleration noise intensity
    r: float - measurement noise standard deviation
    """
    # State Transition Matrix A
    A = np.array([[1, 0, dt, 0],
                  [0, 1, 0, dt],
                  [0, 0, 1, 0],
                  [0, 0, 0, 1]])

    # Measurement Matrix H (position only)
    H = np.array([[1, 0, 0, 0],
                  [0, 1, 0, 0]])

    # Process Noise Covariance Q (White acceleration model)
    G = np.array([[0.5*dt**2, 0], [0, 0.5*dt**2], [dt, 0], [0, dt]])
    Q = q**2 * (G @ G.T)

    # Measurement Noise Covariance R
    R = np.eye(2) * (r**2)

    # Initialization
    x_hat = np.array([Y_obs[0,0], Y_obs[0,1], 0, 0]) # Initial state
    P = np.eye(4) * 100 # Initial large uncertainty

    filtered_states = []

    for y in Y_obs:
        # 1. Prediction
        x_hat = A @ x_hat
        P = A @ P @ A.T + Q

        # 2. Correction
        S = H @ P @ H.T + R
        K = P @ H.T @ np.linalg.inv(S)
        x_hat = x_hat + K @ (y - H @ x_hat)
        P = (np.eye(4) - K @ H) @ P

        filtered_states.append(x_hat.copy())

    return np.array(filtered_states)

# Usage:
filtered_data = filter_gps_measurements(noisy_gps_coords)
```
